```python
from sklearn.feature_extraction.text import CountVectorizer

document = [
    "I love programming in Python",
    "Python is great for data science",
    "Data science is an exciting field",
    "I enjoy learning new programming languages"
]
vectorise = CountVectorizer(ngram_range=(1, 3), stop_words='english')
X = vectorise.fit_transform(document)

print(vectorise.get_feature_names_out())
print(X.toarray())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

document = [
    "I love programming in Python",
    "Python is great for data science",
    "Data science is an exciting field",
    "I enjoy learning new programming languages"
]
vectorise = TfidfVectorizer(ngram_range=(1, 3), stop_words='english')
X = vectorise.fit_transform(document)

print("Voculabulary : ",vectorise.get_feature_names_out())
print("\nBow Matrix : \n",X.toarray())
```
