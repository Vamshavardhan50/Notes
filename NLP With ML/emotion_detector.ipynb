```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
df = pd.read_csv('train.txt',sep=';',header=None,names=['text','emotion'])
df.head()
```

```python
df.isnull().sum()
```

```python
unique_emotions = df['emotion'].unique()
print("Unique emotions:", unique_emotions)
emotion_number = {}
i=0
for emo in unique_emotions:
    emotion_number[emo] = i
    i+=1

df['emotion'] = df['emotion'].map(emotion_number)
```

```python
df.head()
```

```python
df['text'] = df['text'].str.lower()  # Convert text to lowercase
```

```python
import string

def remove_punctuation(text):
    return text.translate(str.maketrans('', '', string.punctuation))
```

```python
df['text'] = df['text'].apply(remove_punctuation)  # Remove punctuation from text
```

```python
df
```

```python
def remove_numbers(text):
    return ''.join([i for i in text if not i.isdigit()])
df['text'] = df['text'].apply(remove_numbers)  # Remove numbers from text
```

```python
def remove_emoji(text):
  new = ""
  for i in text:
    if i.isascii():
      new += i
  return new
df['text'] = df['text'].apply(remove_emoji)  # Remove emojis from text
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
stop_words = set(stopwords.words('english'))
stop_words 
```

```python
df.loc[1]['text']
```

```python
def remove_stopwords(text):
    word_tokens = word_tokenize(text)
    filtered_text = [w for w in word_tokens if not w in stop_words]
    return ' '.join(filtered_text)

df['text'] = df['text'].apply(remove_stopwords)  # Remove stop words from text
```

```python
df
```

```python
df.shape
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df['text'], df['emotion'], test_size=0.2, random_state=42)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer,CountVectorizer
```

```python
bow_vectorizer = CountVectorizer()
x_train_bow = bow_vectorizer.fit_transform(X_train)
x_test_bow = bow_vectorizer.transform(X_test)
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
nb_model = MultinomialNB()
nb_model.fit(x_train_bow, y_train)
```

```python
pred_bow = nb_model.predict(x_test_bow)
```

```python
accuracy_score(y_test, pred_bow)
```

```python
classification_report(y_test, pred_bow)
```

```python
tfid_vector = TfidfVectorizer()
x_train_tfid = tfid_vector.fit_transform(X_train)
x_test_tfid = tfid_vector.transform(X_test)
```

```python
nb2_model = MultinomialNB()
nb2_model.fit(x_train_tfid, y_train)
```

```python
pred_nb2 = nb2_model.predict(x_test_tfid)
```

```python
accuracy_score(y_test, pred_nb2)
```

```python
from sklearn.linear_model import LogisticRegression
log_model = LogisticRegression(max_iter=1000)
log_model.fit(x_train_tfid, y_train)
pred_log = log_model.predict(x_test_tfid)
accuracy_score(y_test, pred_log)
```

```python
import joblib
joblib.dump(log_model, 'logistic_model.pkl')
joblib.dump(tfid_vector, 'tfidf_vectorizer.pkl')
```
